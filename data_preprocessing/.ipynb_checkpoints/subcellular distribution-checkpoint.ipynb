```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sub_Standard_data = pd.read_csv('data/sub_Standard_data.csv')
sub_location = pd.read_excel('data/PPI5093/5093_sub_Standard_data.xlsx', 2, header=None)
sub_location.columns=['subcellular locations']
sub_location
```

```python
# 取出前224个亚细胞位置
sub_location_224 = sub_location.iloc[0:225].values
sub_location_224
```

```python
# 统计每个亚细胞位置包含蛋白质的个数，并降序排序。
protein_counts = sub_Standard_data['sub'].value_counts()  
# 将series转为dataframe
protein_counts_df = {'subcellular locations':protein_counts.index, 'number of proteins':protein_counts.values}
protein_counts_df = pd.DataFrame(protein_counts_df)
protein_counts_df
```

```python
# 左连接
sub_location_pnum = pd.merge(sub_location, protein_counts_df, on=['subcellular locations'], how='left')
sub_location_pnum
```

```python
sub_location_pnum_224=sub_location_pnum[:224]
sub_location_pnum_224
```

```python
sub_loc11 = ["Cytoskeleton", "Cytosol", "Endoplasmic", "Endosome", "Extracellular", "Golgi", "Mitochondrion", "Nucleus", "Peroxisome", "Plasma", "Vacuole"]
```

## 画bar图，不是hist图

```python
fig, ax = plt.subplots(figsize=(6,6))
ax.bar(np.arange(len(sub_location_pnum)),sub_location_pnum['number of proteins'])
ax.set_title('Subcellular Locations Distribution')
ax.set_xlabel('index of subcellular locations')
ax.set_ylabel('number of proteins')
```

```python
fig2, ax2 = plt.subplots(figsize=(6,6))
ax2.bar(np.arange(len(sub_location_pnum_224)),sub_location_pnum_224['number of proteins'])
ax2.set_title('Subcellular Locations Distribution of d Proteins')
ax2.set_xlabel('index of subcellular locations')
ax2.set_ylabel('number of proteins')
```

```python
sub_bool = sub_location_pnum_224['subcellular locations'].isin(sub_loc11)
sub_bool.value_counts()
```

```python
plt.savefig('sub.png')
```
